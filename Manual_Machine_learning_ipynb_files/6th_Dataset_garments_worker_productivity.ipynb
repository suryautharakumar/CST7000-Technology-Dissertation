## Target and Features

```python
target = 'actual_productivity'

X = df.drop(target, axis=1)
y = df[target]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = X.select_dtypes(
    include=['object']
).columns

print("Numerical features:")
print(list(numeric_features))

print("\nCategorical features:")
print(list(categorical_features))
```

```python
#test-train split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## Data Cleaning

```python
X.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer

num_imputer = SimpleImputer(
    strategy='median'
)

X_train[numeric_features] = num_imputer.fit_transform(
    X_train[numeric_features]
)

X_test[numeric_features] = num_imputer.transform(
    X_test[numeric_features]
)
```

```python
cat_imputer = SimpleImputer(
    strategy='most_frequent'
)

X_train[categorical_features] = cat_imputer.fit_transform(
    X_train[categorical_features]
)

X_test[categorical_features] = cat_imputer.transform(
    X_test[categorical_features]
)
```

```python
#check missing values
print("Training missing values:", X_train.isnull().sum().sum())
print("Testing missing values:", X_test.isnull().sum().sum())
```

```python
#convert datetime
X_train['date'] = pd.to_datetime(X_train['date'])
X_test['date'] = pd.to_datetime(X_test['date'])
```

```python
X_train['year'] = X_train['date'].dt.year
X_train['month'] = X_train['date'].dt.month
X_train['day_of_month'] = X_train['date'].dt.day
X_train['day_of_week'] = X_train['date'].dt.dayofweek

X_test['year'] = X_test['date'].dt.year
X_test['month'] = X_test['date'].dt.month
X_test['day_of_month'] = X_test['date'].dt.day
X_test['day_of_week'] = X_test['date'].dt.dayofweek
```

```python
#Remove the original date
X_train = X_train.drop('date', axis=1)
X_test = X_test.drop('date', axis=1)
```

```python
#One-Hot Encoding
X_train = pd.get_dummies(
    X_train,
    drop_first=True
)

X_test = pd.get_dummies(
    X_test,
    drop_first=True
)
```

```python
X_train, X_test = X_train.align(
    X_test,
    join='left',
    axis=1,
    fill_value=0
)

print("Training shape after encoding:", X_train.shape)
print("Testing shape after encoding:", X_test.shape)
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("Feature scaling completed.")
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape)
print("Testing samples:", X_test.shape)
```

## Model Building

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
```

```python
models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree Regressor": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest Regressor": RandomForestRegressor(
        random_state=42
    ),

    "Gradient Boosting Regressor": GradientBoostingRegressor(
        random_state=42
    )
}
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np

results = []

for name, model in models.items():

    # Train model
    model.fit(X_train, y_train)

    # Make predictions
    y_pred = model.predict(X_test)

    # MAE
    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    # RMSE
    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred
        )
    )

    # R² Score
    r2 = r2_score(
        y_test,
        y_pred
    )

    results.append([
        name,
        mae,
        rmse,
        r2
    ])
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "MAE",
        "RMSE",
        "R² Score"
    ]
)

# Sort by R² Score
results_df = results_df.sort_values(
    by="R² Score",
    ascending=False
).reset_index(drop=True)

results_df
```

```python
best_model = results_df.iloc[0]

print("Best Performing Model")
print("----------------------")
print("Model:", best_model["Model"])
print("MAE:", round(best_model["MAE"], 6))
print("RMSE:", round(best_model["RMSE"], 6))
print("R² Score:", round(best_model["R² Score"], 6))
```

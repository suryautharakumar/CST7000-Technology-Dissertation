## Define Target and Features

```python
X = df.drop('actual_productivity', axis=1)
y = df['actual_productivity']

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

## Data Cleaning

```python
X.isnull().sum()
```

```python
from sklearn.compose import make_column_selector

numeric_features = X.select_dtypes(include=['int64','float64']).columns

categorical_features = X.select_dtypes(include=['object']).columns

print("Numerical Columns")
print(numeric_features)

print("\nCategorical Columns")
print(categorical_features)
```

```python
from sklearn.impute import SimpleImputer

#Numerical columns
num_imputer = SimpleImputer(strategy='median')
X[numeric_features] = num_imputer.fit_transform(X[numeric_features])

#Categorical columns
cat_imputer = SimpleImputer(strategy='most_frequent')
X[categorical_features] = cat_imputer.fit_transform(X[categorical_features])

print("Missing values after imputation")
print(X.isnull().sum())
```

```python
#One-Hot Encoding
X = pd.get_dummies(X, drop_first=True)

print("Shape after encoding:", X.shape)
```

```python
#Feature Scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(X)
```

```python
#Train-Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training samples:", X_train.shape)
print("Testing samples:", X_test.shape)
```

## Model Building

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Linear Regression Performance")
print("------------------------------")
print("MAE :", round(mae,4))
print("MSE :", round(mse,4))
print("RMSE:", round(rmse,4))
print("R² Score:", round(r2,4))
```

## Compare Multiple Regression Models

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree Regressor": DecisionTreeRegressor(random_state=42),
    "Random Forest Regressor": RandomForestRegressor(random_state=42),
    "Gradient Boosting Regressor": GradientBoostingRegressor(random_state=42)
}
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import pandas as pd
import numpy as np

results = []

for name, model in models.items():

    # Train model
    model.fit(X_train, y_train)

    # Predict
    y_pred = model.predict(X_test)

    # Evaluation
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    results.append([name, mae, rmse, r2])

# Create comparison table
results_df = pd.DataFrame(results, columns=[
    "Model",
    "MAE",
    "RMSE",
    "R² Score"
])

results_df.sort_values(by="R² Score", ascending=False).reset_index(drop=True)
```

```python
best_model = results_df.sort_values(by="R² Score", ascending=False).iloc[0]

print("Best Performing Model")
print("----------------------")
print(best_model)
```

## Data Cleaning

```python
X.isnull().sum()
```

```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
num_imputer = SimpleImputer(strategy='median')

X_train[numeric_features] = num_imputer.fit_transform(
    X_train[numeric_features]
)

X_test[numeric_features] = num_imputer.transform(
    X_test[numeric_features]
)
```

```python
if len(categorical_features) > 0:

    cat_imputer = SimpleImputer(
        strategy='most_frequent'
    )

    X_train[categorical_features] = cat_imputer.fit_transform(
        X_train[categorical_features]
    )

    X_test[categorical_features] = cat_imputer.transform(
        X_test[categorical_features]
    )
```

```python
#One-Hot Encoding

X_train = pd.get_dummies(
    X_train,
    drop_first=True
)

X_test = pd.get_dummies(
    X_test,
    drop_first=True
)

X_train, X_test = X_train.align(
    X_test,
    join='left',
    axis=1,
    fill_value=0
)

print("Training shape after encoding:", X_train.shape)
print("Testing shape after encoding:", X_test.shape)
```

```python
#Feature Scaling

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("Feature scaling completed.")
```

## Model Building

```python
models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000
    ),

    "Decision Tree Classifier": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest Classifier": RandomForestClassifier(
        random_state=42
    ),

    "Gradient Boosting Classifier": GradientBoostingClassifier(
        random_state=42
    )
}
```

```python
results = []

for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    y_prob = model.predict_proba(X_test)

    positive_class_index = list(model.classes_).index(
        y.unique()[1]
    )

    y_prob_positive = y_prob[:, positive_class_index]

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        average='weighted',
        zero_division=0
    )

    recall = recall_score(
        y_test,
        y_pred,
        average='weighted',
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        y_pred,
        average='weighted',
        zero_division=0
    )

    # ROC-AUC
    if len(model.classes_) == 2:

        roc_auc = roc_auc_score(
            y_test,
            y_prob_positive
        )

    else:

        roc_auc = roc_auc_score(
            y_test,
            y_prob,
            multi_class='ovr',
            average='weighted'
        )

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc
    ])
```

```python
results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC-AUC"
    ]
)

results_df = results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

results_df
```

```python
best_model = results_df.iloc[0]

print("Best Performing Model")
print("----------------------")
print("Model:", best_model["Model"])
print("Accuracy:", round(best_model["Accuracy"], 4))
print("Precision:", round(best_model["Precision"], 4))
print("Recall:", round(best_model["Recall"], 4))
print("F1 Score:", round(best_model["F1 Score"], 4))
print("ROC-AUC:", round(best_model["ROC-AUC"], 4))
```
